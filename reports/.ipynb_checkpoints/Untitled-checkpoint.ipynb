```python
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt
import math

prefix = "../results/"
# postfix="_00_demo.dat"
postfix = "_01_big_query.dat"
# postfix = "_02_index.dat"
# postfix = "_03_select.dat"

f = prefix + "postgres" + postfix
# f = prefix + "pg_duckdb" + postfix
# f = prefix + "pg_duckdb_parquet" + postfix
t = []
cnt=0
with open(f, 'r') as file:
    for line in file:
        cnt += 1
        clear = line.strip()
        if clear:
            clear_dot = clear.replace(',', '.')
            t.append(float(clear_dot))
print(t)
plt.hist(t)
test_normal = stats.normaltest(t)
test_shapiro = stats.shapiro(t)
print(test_normal, test_shapiro, sep = '\n')
if test_normal[1] > 0.05 or test_shapiro[1] > 0.05:
    print("Тест пройден")
else:
    print("!> ТЕСТ НЕ ПРОЙДЕН")
mean = np.mean(t)
print(mean)
std = np.std(t, ddof=1)
print(std)
quality = (std / mean) * 100
print(quality, "%")

```
